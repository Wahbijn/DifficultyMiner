```python
import requests
import csv

# -------------------
# 1️⃣ CHOISIR LE JEU
# -------------------

APPIDS = [
    570, # Dota 2
    730, # CS:GO
    440, # Team Fortress 2
    2668510, # RED DEAD REDEMPTION 2
    3240220, # GTA V
    1245620, # Elden Ring
    1551360, # FORZA HORIZON 5
    578080, # PUBG: BATTLEGROUNDS
    3764200, # RESIDENT EVIL Requiem
    2358720, # BLACK MYTH WUKONG
    311210 # CALL OF DUTY®: Black Ops III
]


# -------------------
# 2️⃣ RÉCUPÉRER NOM JEU
# -------------------


rows = []

for APPID in APPIDS:

    print("Scraping", APPID)

    # -------- metadata --------
    meta_url = f"https://store.steampowered.com/api/appdetails?appids={APPID}"
    meta = requests.get(meta_url).json()

    if not meta[str(APPID)]["success"]:
        print("skip — no data")
        continue

    game_name = meta[str(APPID)]["data"]["name"]

    # -------------------
    # 3️⃣ API REVIEWS
    # -------------------

    review_url = f"https://store.steampowered.com/appreviews/{APPID}"

    params = {
        "json": 1,
        "num_per_page": 100
    }

    

    data = requests.get(review_url, params=params).json()

    # -------------------
    # 4️⃣ EXTRACTION FEATURES
    # -------------------


    for r in data["reviews"]:

        text = r["review"]

        rows.append([
            APPID,
            game_name,
            r["author"]["steamid"],
            len(text),
            len(text.split()),
            r["votes_up"],
            r["votes_funny"],
            r["voted_up"],
            r["author"]["playtime_at_review"],
            text.replace("\n"," ")
        ])

# -------- save --------

with open("reviews.csv", "w", newline="", encoding="utf-8-sig") as f:

    writer = csv.writer(f)

    writer.writerow([
        "game_id",
        "game_name",
        "steamid",
        "review_len_chars",
        "review_len_words",
        "votes_helpful",
        "votes_funny",
        "recommended",
        "playtime",
        "review_text"
    ])

    writer.writerows(rows)

print("✅ ALL GAMES DONE")
```
